_Bagian ini dikutip dari buku PDF (sumber kebenaran praktikum) dan diformat ulang agar mudah dibaca. Kode asli buku tidak diubah; bagian instalasi/persiapan lingkungan diringkas karena sudah dijelaskan di README._

### 4.1. Update sistem terlebih dahulu

Buka terminal, lalu jalankan:

```python
sudo apt update
sudo apt upgrade -y
```

Langkah ini penting agar indeks paket dan dependensi sistem berada dalam kondisi terbaru.

### 4.2. Pastikan Python dan tool dasar tersedia

python3 --version pip3 --version

Bila pip3 belum tersedia, pasang dengan:

```python
sudo apt install -y python3-pip python3-venv
```

Penjelasan:

- python3-pip dipakai untuk memasang pustaka Python
- python3-venv dipakai untuk membuat virtual environment

### 4.3. Buat folder kerja utama

Misalnya kita akan menaruh semua proyek buku di folder iot-ml-book pada home directory:

mkdir -p ~/iot-ml-book cd ~/iot-ml-book

Cek isi folder:

pwd ls -la

### 4.4. Buat virtual environment

python3 -m venv .venv

Aktifkan:

source .venv/bin/activate

Kalau berhasil, prompt terminal biasanya berubah, misalnya menjadi:

```python
(.venv) user@ubuntu:~/iot-ml-book$
```

Kenapa ini penting? Karena lingkungan ini akan mengisolasi paket proyek dari paket sistem. Jadi instalasi untuk buku ini tidak “mengotori” Python global Ubuntu.

### 4.5. Upgrade pip dan pasang pustaka inti

```python
pip install --upgrade pip
pip install jupyterlab pandas numpy matplotlib scipy
scikit-learn
```

Untuk praktik awal Bab 1, paket di atas sudah cukup.

Fungsi masing-masing

- jupyterlab: lingkungan kerja interaktif
- pandas: membaca dan mengolah CSV/tabular data
- numpy: operasi numerik
- matplotlib: visualisasi dasar

- scipy: fungsi ilmiah pendukung
- scikit-learn: dipakai di bab ML berikutnya

### 4.6. Simpan daftar paket

Agar proyek mudah direplikasi di komputer lain:

```python
pip freeze > requirements.txt
```

Cek hasil:

cat requirements.txt

Ini kebiasaan penting dalam proyek profesional.

### 4.7. Jalankan JupyterLab

```python
jupyter lab
```

Biasanya terminal akan menampilkan URL lokal. Buka URL tersebut di browser. Dari sini pembaca bisa membuat:

- notebook baru
- file Python
- folder data
- dokumentasi eksperimen

### 5. Struktur folder proyek yang disarankan

Sejak awal, biasakan struktur proyek seperti ini:

iot-ml-book/ ├── .venv/ ├── data/ │   ├── raw/ │   └── processed/ ├── notebooks/ ├── scripts/ ├── outputs/ │   ├── figures/ │   └── reports/ ├── requirements.txt └── README.md

Buat foldernya:

mkdir -p data/raw data/processed notebooks scripts outputs/figures outputs/reports touch README.md

Arti masing-masing folder

- data/raw/ → data mentah asli
- data/processed/ → data hasil pembersihan
- notebooks/ → eksperimen Jupyter
- scripts/ → skrip Python yang siap dijalankan ulang
- outputs/figures/ → hasil grafik
- outputs/reports/ → laporan, ringkasan, atau ekspor hasil

Struktur seperti ini akan sangat membantu saat proyek berkembang menjadi pipeline MQTT, dashboard, dan model ML.
